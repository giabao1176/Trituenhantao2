Bài toán sẽ tạo ra ngẫu nhiên một ma trận có kích thước m x n theo người dùng nhập, sau đó sẽ tạo ngẫu nhiên các ô bẩn hay không ($1$: "bẩn", $0$: "sạch") và robot thực hiện bằng cách sẽ kiểm tra các ô xung quanh nó, nếu bẩn thì sẽ đi qua nếu sạch thì sẽ thực hiện Random ngẫu nhiên bước đi và sau đó tính lại vị trí hiện tại và tiếp tục kiểm tra các ô xung quanh nó để tiếp tục di chuyển và sẽ dừng lại khi tất cả các ô kiểm tra hội tụ (là một ma trận toàn $0$) hoặc sẽ dừng khi đạt số epoch là 50.

```python
import random

def hien_thi(ma_tran, r, c):
    for i in range(len(ma_tran)):
        row = []
        for j in range(len(ma_tran[0])):
            if i == r and j == c:
                row.append(f"[{ma_tran[i][j]}]")
            else:
                row.append(f" {ma_tran[i][j]} ")
        print(" ".join(row))
    print("-" * 15)

def robot_hut_bui(m, n):
    ma_tran = [[random.randint(0, 1) for _ in range(n)] for _ in range(m)]
    r, c = random.randint(0, m-1), random.randint(0, n-1)
    
    for buoc in range(50):
        if ma_tran[r][c] == 1:
            ma_tran[r][c] = 0
        
        print(f"Step: {buoc + 1}")
        hien_thi(ma_tran, r, c)

        if not any(1 in row for row in ma_tran):
            print("Done!")
            break

        huong = []
        if r > 0: huong.append((-1, 0))
        if r < m - 1: huong.append((1, 0))
        if c > 0: huong.append((0, -1))
        if c < n - 1: huong.append((0, 1))

        o_ban = [(dr, dc) for dr, dc in huong if ma_tran[r+dr][c+dc] == 1]
        
        dr, dc = random.choice(o_ban if o_ban else huong)
        r, c = r + dr, c + dc

m = int(input("M: "))
n = int(input("N: "))
robot_hut_bui(m, n)
```
