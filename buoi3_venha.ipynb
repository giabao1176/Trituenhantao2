Bài toán sẽ tạo ra ngẫu nhiên một ma trận có kích thước m x n theo người dùng nhập, sau đó sẽ tạo ngẫu nhiên các ô bẩn hay không ($1$: "bẩn", $0$: "sạch") và các vật cản ($-1$: "vật cản"). Robot thực hiện bằng cách sẽ kiểm tra các ô xung quanh nó, nếu bẩn thì sẽ đi qua nếu sạch thì sẽ thực hiện Random ngẫu nhiên bước đi (không đi vào vật cản) và sẽ dừng lại khi tất cả các ô bẩn đã được làm sạch hoặc sẽ dừng khi đạt số bước là 50(nên nhiều trường hợp sẽ không thể hội tụ vì số epoch không quá nhiều "50").

```python
import random

def hienthi(matran, r, c):
    for i in range(len(matran)):
        row = []
        for j in range(len(matran[0])):
            if i == r and j == c:
                row.append(f"[{matran[i][j]}]")
            elif matran[i][j] == -1:
                row.append(" # ") # Tương đương vật cản (-1)
            else:
                row.append(f" {matran[i][j]} ")
        print(" ".join(row))


def robot(m, n, sovatcan):
    # Khởi tạo ma trận
    matran = [[random.randint(0, 1) for _ in range(n)] for _ in range(m)]
    
    # Thêm vật cản (-1)
    count = 0
    while count < sovatcan:
        vr, vc = random.randint(0, m-1), random.randint(0, n-1)
        if matran[vr][vc] != -1:
            matran[vr][vc] = -1
            count += 1
            
    # Vị trí xuất phát của robot
    while True:
        r, c = random.randint(0, m-1), random.randint(0, n-1)
        if matran[r][c] != -1:
            break
    
    print("Trạng thái ban đầu")
    hien_thi(matran, r, c)

    for buoc in range(1, 51):
        # Nếu ô hiện tại bẩn thì làm sạch
        if matran[r][c] == 1:
            matran[r][c] = 0
        
        print(f"Step: {buoc}")
        hien_thi(matran, r, c)

        # Kiểm tra nếu hết rác
        if not any(1 in row for row in matran):
            print("Xong.")
            break

        # Tìm các hướng đi hợp lệ (không ra ngoài biên và không đâm vào vật cản)
        huong_hop_le = []
        for dr, dc in [(-1, 0), (1, 0), (0, -1), (0, 1)]:
            nr, nc = r + dr, c + dc
            if 0 <= nr < m and 0 <= nc < n and matran[nr][nc] != -1:
                huong_hop_le.append((dr, dc))
        
        if not huong_hop_le:
            print("Robot bị kẹt!")
            for s in range(buoc + 1, 51):
                print(f"Step: {s} (Stuck)")
            break

        # Ưu tiên đi vào ô bẩn xung quanh
        o_ban = [(dr, dc) for dr, dc in huong_hop_le if matran[r+dr][c+dc] == 1]
        
        dr, dc = random.choice(o_ban if o_ban else huong_hop_le)
        r, c = r + dr, c + dc

    print("\nKết quả cuối cùng")
    hien_thi(matran, r, c)

m = int(input("Nhập số hàng M: "))
n = int(input("Nhập số cột N: "))
s = int(input("Nhập số vật cản: "))
robot(m, n, s)
```
