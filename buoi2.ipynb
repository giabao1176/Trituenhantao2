```python
from collections import deque

def tim_o_trong(board):
    for r in range(3):
        for c in range(3):
            if board[r][c] == 0:
                return r, c

def lay_hanh_dong(r, c):
    actions = []
    if r > 0: actions.append('U')
    if r < 2: actions.append('D')
    if c > 0: actions.append('L')
    if c < 2: actions.append('R')
    return actions

def di_chuyen(board, action):
    new_board = [row[:] for row in board]
    r, c = tim_o_trong(new_board)
    if action == 'U': new_board[r][c], new_board[r-1][c] = new_board[r-1][c], new_board[r][c]
    if action == 'D': new_board[r][c], new_board[r+1][c] = new_board[r+1][c], new_board[r][c]
    if action == 'L': new_board[r][c], new_board[r][c-1] = new_board[r][c-1], new_board[r][c]
    if action == 'R': new_board[r][c], new_board[r][c+1] = new_board[r][c+1], new_board[r][c]
    return new_board

def solve(start, goal):
    queue = deque([(start, [])])
    visited = {tuple(tuple(row) for row in start)}
    
    while queue:
        current_board, path = queue.popleft()
        if current_board == goal:
            return path
        
        r, c = tim_o_trong(current_board)
        for action in lay_hanh_dong(r, c):
            next_board = di_chuyen(current_board, action)
            state = tuple(tuple(row) for row in next_board)
            if state not in visited:
                visited.add(state)
                queue.append((next_board, path + [action]))
    return None

start_board = [
    [1, 2, 3],
    [4, 0, 5],
    [6, 7, 8]
]

goal_board = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 0]
]

print("Trạng thái bắt đầu:")
for row in start_board: print(row)

path = solve(start_board, goal_board)

if path:
    current = start_board
    for i, action in enumerate(path):
        current = di_chuyen(current, action)
        print(f"\nBước {i+1} - Di chuyển '{action}':")
        for row in current: print(row)
else:
    print("\nKhông tìm thấy lời giải.")
```
